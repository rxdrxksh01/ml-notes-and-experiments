```python
import numpy as np 
import pandas as pd 
```

```python
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer,MissingIndicator
```

```python
df = pd.read_csv('../sample_data/train.csv',usecols=['Age','Fare','Survived'])
```

```python
df.head()
```

```python
x = df.drop('Survived',axis=1)
```

```python
y = df['Survived']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42 )
```

```python

si = SimpleImputer()
x_train_trans = si.fit_transform(x_train)
x_test_trans=si.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
lr.fit(x_train_trans,y_train)
y_pred = lr.predict(x_test_trans)

from sklearn.metrics import accuracy_score

accuracy_score(y_test,y_pred)
```

```python
mi = MissingIndicator()
x_train_t = mi.fit_transform(x_train)
x_test_t = mi.transform(x_test)

```

```python
x_train['age_na'] = x_train_t
x_test['age_na'] = x_test_t
```

```python
sip = SimpleImputer()
x_train_trf = sip.fit_transform(x_train)
x_test_trf = sip.transform(x_test)

```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
lr.fit(x_train_trf,y_train)
y_pred = lr.predict(x_test_trf)

from sklearn.metrics import accuracy_score

accuracy_score(y_test,y_pred)
```

```python
si_imp = SimpleImputer(add_indicator=True)
x_training = si_imp.fit_transform(x_train)
x_testing = si_imp.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
lr.fit(x_training,y_train)
y_pred = lr.predict(x_testing)

from sklearn.metrics import accuracy_score

accuracy_score(y_test,y_pred)
```
